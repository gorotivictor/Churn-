```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from imblearn.over_sampling import SMOTE
from sklearn.preprocessing import MinMaxScaler
import xgboost as xgb
import joblib
import seaborn as sns
import plotly.graph_objects as go
sns.set(font_scale=2)
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1. Load Synthetic Transactional Data

```python
bank_name = "Sunrise Microfinance Bank"
df = pd.read_csv('synthetic_data_sunrise_microfinance_bank.csv')
```

```python
# 2. Data Preprocessing and Feature Engineering
```

```python
#Check for missing values
df.isnull().sum()
```

```python
df['CustomerId'].value_counts(ascending=False)
```

```python
# Drop non-predictive columns
df = df.drop(['CustomerId', 'Surname'], axis=1)
```

```python

# 2. Feature Encoding
# Label Encoding for binary/ordinal categorical variables
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
df['Account Activity'] = df['Account Activity'].map({'Active': 0, 'Dormant': 1})
df['Repayment Timeliness'] = df['Repayment Timeliness'].map({'On-time': 0, 'Late': 1})
df['Use of Savings Products'] = df['Use of Savings Products'].astype(int)
df['Use of Loan Products'] = df['Use of Loan Products'].astype(int)
df['Participation in Group Lending'] = df['Participation in Group Lending'].astype(int)

df['Account Balance Trend'] = df['Account Balance Trend'].map({
    'Decreasing': 0,
    'Stable': 1,
    'Increasing': 2
})

# One-Hot Encoding for non-ordinal categorical variables
df = pd.get_dummies(df, columns=['Marital Status', 'Education Level', 'Loan History', 'Use of Digital Banking'], prefix=['Marital', 'Education', 'Loan_History', 'Digital_Banking'])
```

```python
# 3. Feature Scaling
scaler = MinMaxScaler()
numeric_columns = [
    'Age', 'Tenure', 'Frequency of Deposits/Withdrawals', 'Average Transaction Value',
    'Number of Complaints Logged', 'Response Time to Complaints', 'Customer Support Interactions',
    'Overdue Loan Frequency', 'Penalties Paid'
]
df[numeric_columns] = scaler.fit_transform(df[numeric_columns])
```

```python
df
```

```python
# Features and target
X = df[['Age', 'Gender', 'Tenure', 'Frequency of Deposits/Withdrawals', 'Repayment Timeliness', 
        'Account Activity', 'Account Balance Trend']]
y = df['Exited']
```

```python
# Handle imbalanced data using SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
```

```python
# 3. Split data
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.3, random_state=42)
```

```python
# Plot class distribution
plt.figure(figsize=(12, 5))
sns.countplot(x=y_train, color='darkred')
plt.title('Exited Distribution')
plt.show()
print("\nClass Distribution in Training Set:")
print(y_train.value_counts(normalize=True))
```

```python
X
```

```python
# Correlation matrix for numeric features
numeric_columns = [
    'Age', 'Gender', 'Tenure', 'Frequency of Deposits/Withdrawals', 'Repayment Timeliness', 
        'Account Activity', 'Account Balance Trend'
]
corr = X_train[numeric_columns].corr()
plt.figure(figsize=(25, 12))
mask = np.tril(np.ones_like(corr))
sns.heatmap(corr, annot=True, fmt=".2f", mask=mask, square=True)
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```python
# 4. Model Development
```

```python
models = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42, n_estimators=100),
    'XGBoost': xgb.XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss')
}
```

```python
# Train and evaluate models
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_pred_proba = model.predict_proba(X_test)[:, 1]
    results[name] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1-Score': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_pred_proba)
    }
```

```python
# Print results
print('Model Evaluation Results:')
for name, metrics in results.items():
    print(f'\n{name}:')
    for metric, value in metrics.items():
        print(f'{metric}: {value:.4f}')
```

```python
# Select best model (based on F1-Score)
best_model_name = max(results, key=lambda x: results[x]['F1-Score'])
best_model = models[best_model_name]
print(f'\nBest Model: {best_model_name}')
```

```python
# Save best model
joblib.dump(best_model, 'best_model.pkl')
```

```python
# Feature importance (for Random Forest or XGBoost)
if best_model_name in ['Random Forest', 'XGBoost']:
    feature_importance = pd.DataFrame({
        'Feature': X.columns,
        'Importance': best_model.feature_importances_
    }).sort_values(by='Importance', ascending=False)
    print('\nFeature Importance:')
    print(feature_importance)
    print('\nRecommendations:')
    print('- Focus on reducing failed transactions to retain customers.')
    print('- Monitor dormant accounts and engage customers with low transaction frequency.')
    print('- Improve customer support response times to reduce interaction intensity.')
```
